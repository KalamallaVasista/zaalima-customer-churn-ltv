```python
categorical_columns = [
    "Contract",
    "InternetService",
    "PaymentMethod",
    "TechSupport",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "StreamingTV",
    "StreamingMovies"
]

eda_results = []

for column in categorical_columns:
    churn_rate = (
        df.groupby(column)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
    )

    for category, rate in churn_rate.items():
        eda_results.append({
            "Feature": column,
            "Category": category,
            "ChurnRate": round(rate, 2)
        })

eda_summary = pd.DataFrame(eda_results)

eda_summary.sort_values(
    "ChurnRate",
    ascending=False
).head(20)
```

```python
correlation_df = df.copy()

correlation_df["Churn_numeric"] = (
    correlation_df["Churn"]
    .map({"No": 0, "Yes": 1})
)

correlation_df["TotalCharges_numeric"] = pd.to_numeric(
    correlation_df["TotalCharges"],
    errors="coerce"
)

numeric_columns = [
    "SeniorCitizen",
    "tenure",
    "MonthlyCharges",
    "TotalCharges_numeric",
    "Churn_numeric"
]

correlation_matrix = correlation_df[numeric_columns].corr()

print(correlation_matrix.round(2))
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Matrix of Numerical Variables")

plt.show()
```
